This notebook is the **modern, CI-tested entry point** for the repository. It runs Thai Named Entity Recognition without requiring the LST20 dataset.

**2026 quickstart choice:** this notebook intentionally uses PyThaiNLP's lightweight `thainer` CRF engine so the entry point stays fast and CI-friendly. The maintained WangchanBERTa path is tested separately under `src/thai_ner_2026/`.

For the maintained WangchanBERTa + LST20 training path, see `notebooks/10_wangchanberta_lst20_2026.ipynb`.

```python
from pythainlp.tag import NER

# Lightweight PyThaiNLP CRF baseline for the 2026 quickstart.
ner = NER("thainer")
```

## Run NER on a Thai sentence

The example below contains person, organization, and location-like information so the output is easy to inspect.

```python
text = "นายสมชายทำงานที่ธนาคารแห่งหนึ่งในกรุงเทพมหานคร"
entities = ner.tag(text, pos=False)
entities
```

```python
assert isinstance(entities, list)
assert len(entities) > 0
assert all(isinstance(item, tuple) and len(item) == 2 for item in entities)
entity_labels = sorted({label for _, label in entities if label != "O"})
print(f"NER produced {len(entities)} tagged tokens.")
print(f"Detected entity labels: {entity_labels or 'none'}")
```
